### Display the Atlas Brain graphs as IPython flowcharts

```python
from IPython.display import Image, display

from Organs.brain import Brain


def display_flowchart(title: str, graph) -> None:
    """Render one compiled LangGraph graph as an embedded PNG image."""
    print(title)
    display(Image(data=graph.get_graph().draw_mermaid_png()))


def display_all_flowcharts() -> None:
    """Display the three Brain subgraphs followed by the master graph."""
    brain = Brain()

    flowcharts = (
        ("Graph 1: Input Preparation and LLM Selection", brain.ready_for_processing()),
        ("Graph 2: LLM Processing", brain._build_processing_graph()),
        ("Graph 3: Sanitization and Mouth Output", brain.ready_to_speak()),
        ("Master Graph: Complete Cognitive Pipeline", brain.graph),
    )

    for title, graph in flowcharts:
        display_flowchart(title, graph)


display_all_flowcharts()
```
